```python
from flask import Flask, request, jsonify
from flask_cors import CORS
import time
from pyngrok import ngrok
from pinecone import Pinecone, ServerlessSpec
from semantic_router.encoders import HuggingFaceEncoder
import requests
from groq import Groq
import json
import threading

app = Flask(__name__)
CORS(app)  # Apply CORS to the Flask app

# Define API keys and index mappings
API_KEYS = {
    "default": "pcsk_4DMnCt_G3q8k4fqMST5bBg1XPQs9KwmC7H9jNUwMsxAe6hytxFz8zmM6ftjTVv6R7yBzXG",  # Replace with actual default API key
    "hs_am": "pcsk_6a3QNU_Tu86RkpxTrkCJXQCbSnNrP2YPNHrya8oCVXobQ5dti7i7FcaoirMEuCwoxynxWa",  # Replace with actual HS/AM API key
    "uni": "",  # Replace with actual UNI API key (if available)
}

INDEX_MAPPING = {
    "3HS": "3hs-history",
    "2HS": "2hs-history",
    "1HS": "1hs-history",
    "1AM": "1am-history",
    "2AM": "2am-history",
    "3AM": "3am-history",
    "4AM": "4am-history",
    "3AP": "3ap-history",
    "4AP": "4ap-history",
    "5AP": "5ap-history",
    "UNI": "uni-history",
}

# Pinecone ServerlessSpec configuration
SPEC = ServerlessSpec(cloud="aws", region="us-east-1")


def get_api_key(level):
    """Get the API key based on the level."""
    if level in ["1HS", "2HS", "3HS", "1AM", "4AM"]:
        return API_KEYS["hs_am"]
    elif level == "UNI":
        return API_KEYS["uni"]
    else:
        return API_KEYS["default"]


def initialize_pinecone_index(level):
    """
    Initialize Pinecone index based on the specified level.

    Args:
        level (str): The level for which the index should be initialized.

    Returns:
        Pinecone.Index: The initialized Pinecone index object.
    """
    index_name = INDEX_MAPPING.get(level)
    if not index_name:
        raise ValueError(f"Invalid level: {level}. Available levels are {list(INDEX_MAPPING.keys())}")

    api_key = get_api_key(level)
    if not api_key:
        raise ValueError(f"No API key available for level: {level}")

    # Initialize Pinecone connection
    pc = Pinecone(api_key=api_key)

    # Check if the index exists and create it if necessary
    existing_indexes = [index_info["name"] for index_info in pc.list_indexes()]
    if index_name not in existing_indexes:
        pc.create_index(index_name=index_name, dimension=768, metric='cosine', spec=SPEC)
        while not pc.describe_index(index_name).status['ready']:
            time.sleep(1)

    # Connect to the index
    return pc.Index(index_name)


# Initialize encoder
print("Initializing encoder...")
try:
    encoder = HuggingFaceEncoder(name="gimmeursocks/ara-e5-base")
    print("Encoder initialized successfully.")
except Exception as e:
    print(f"Error initializing encoder: {e}")
    raise
def parse_response(output):
    # Ensure the output is a string
    if isinstance(output, list):
        output = output[0]

    # Define the start and end markers
    start_marker = "### Response:\n"
    end_marker = "<|eot_id|>"

    # Extract the response between the markers
    if start_marker in output and end_marker in output:
        start_index = output.index(start_marker) + len(start_marker)
        end_index = output.index(end_marker)
        response = output[start_index:end_index].strip()
        return response
    return None

alpaca_prompt = """Below is an instruction that describes a task, paired with an input that provides further context. Write a response that appropriately completes the request.

### Instruction:
Answer the following question in Arabic about History:

### Input:
{}

### Response:
{}"""

def get_history_answers(question, alpaca_prompt=alpaca_prompt, model=history_model, tokenizer=history_tokenizer):

    # Prepare the input using the tokenizer
    inputs = tokenizer(
        [
            alpaca_prompt.format(
                question,  # Question is passed here
                "",        # Output - leave this blank for generation
            )
        ], return_tensors="pt"
    ).to("cuda")

    outputs = model.generate(**inputs, max_new_tokens = 128, temperature=0.1, use_cache = True)
    tokenizer.batch_decode(outputs)
    return parse_response(tokenizer.batch_decode(outputs))


def get_docs(query: str, top_k: int, level: str) -> list:
    """Retrieve documents from the Pinecone index."""
    try:
        print(f"Retrieving documents for query: {query} with top_k: {top_k} and level: {level}")
        index = initialize_pinecone_index(level)  # Initialize the appropriate index
        xq = encoder([query])
        res = index.query(vector=xq, top_k=top_k, include_metadata=True)
        docs = [x["metadata"]["content"] for x in res["matches"]]
        print(f"Retrieved documents: {docs}")
        return docs
    except Exception as e:
        print(f"Error retrieving documents: {e}")
        raise

history_tool = {
    "type": "function",
    "function": {
        "name": "get_history_answers",
        "description": "Provides historical information and answers to history queries.",
        "parameters": {
            "type": "object",
            "properties": {
                "question": {
                    "type": "string",
                    "description": "The historical question or topic needing an answer."
                }
            },
            "required": ["question"]
        }
    }
}

client = Groq(api_key="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")
MODEL = "llama-3.2-90b-vision-preview"

@app.route('/query', methods=['POST'])
def query():
    try:
        print("Received /query POST request")
        data = request.json
        print(f"Request data: {data}")

        question = data.get('question')
        level = data.get('level')
        if not question:
            print("No question provided in the request.")
            return jsonify({"error": "Question is required"}), 400

        if not level:
            print("No level provided in the request.")
            return jsonify({"error": "Level is required"}), 400

        # Step 1: Retrieve documents from Pinecone
        try:
            docs = get_docs(query=question, top_k=2, level=level)
            if not docs:
                print("No relevant documents found.")
                return jsonify({"error": "No relevant documents found."}), 404
            context = "\n---\n".join(docs)
            print(f"Retrieved context: {context}")
        except Exception as e:
            print(f"Error retrieving documents: {e}")
            return jsonify({"error": f"Error retrieving documents: {str(e)}"}), 500

        # Step 2: Build system message with context
        system_message = (
            "You are a helpful assistant that answers questions about Algerian History for students only in Arabic "
            "this context can be useful but do not rely only on it \n\n"
            f"CONTEXT:\n{context}"
        )
        history = [
            {
                "role": "system",
                "content": system_message
            }
        ]

        # Step 3: Check for "exit" condition
        if question.lower() == "exit":
            response_message = "Ending conversation. Goodbye!"
            history.append({"role": "user", "content": response_message})
            return jsonify({
                "response": response_message,
                "docs": docs
            }), 200

        # Step 4: Add user input to conversation history
        history.append({"role": "user", "content": question})

        # Step 5: Process conversation using Groq model
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=history,
                stream=False,
                tools=[history_tool],
                tool_choice="auto",
                temperature=0,
                max_tokens=4096
            )
            response_message = response.choices[0].message
            print(f"Initial response: {response_message.content}")

            # Append assistant's initial response
            history.append({"role": "assistant", "content": response_message.content})

            # Handle tool calls (if any)
            tool_calls = response_message.tool_calls
            if tool_calls:
                available_functions = {"get_history_answers": get_history_answers}

                for tool_call in tool_calls:
                    try:
                        function_name = tool_call.function.name
                        function_to_call = available_functions.get(function_name)

                        if function_to_call:
                            function_args = json.loads(tool_call.function.arguments)
                            function_response = function_to_call(
                                question=function_args.get("question")
                            )
                            history.append({
                                "role": "tool",
                                "name": function_name,
                                "content": function_response,
                                "tool_call_id": tool_call.id
                            })
                            print(f"Tool '{function_name}' response: {function_response}")
                    except Exception as e:
                        print(f"Error in tool '{tool_call.function.name}': {e}")

            # Step 6: Re-run the model with updated history
            final_response = client.chat.completions.create(
                model=MODEL,
                temperature=0,
                messages=history
            )
            assistant_response = final_response.choices[0].message.content
            print(f"Final assistant response: {assistant_response}")

            # Update conversation history
            history.append({"role": "assistant", "content": assistant_response})

            # Return response
            return jsonify({
                "response": assistant_response,
                "docs": docs
            }), 200

        except Exception as e:
            print(f"Error processing conversation: {e}")
            return jsonify({"error": f"Error processing conversation: {str(e)}"}), 500

    except Exception as e:
        print(f"Error in /query: {e}")
        return jsonify({"error": "Internal server error"}), 500

# Start Flask app in a separate thread
def run_flask():
    app.run(host='0.0.0.0', port=5000)

# Start ngrok tunnel
def start_ngrok():
    public_url = ngrok.connect(5000)
    print(f"ngrok is running! Public URL: {public_url}")

# Run Flask and ngrok in parallel
threading.Thread(target=run_flask).start()
start_ngrok()
```
